```python

# STEP 3: MERGE ALL DATASETS
# ========================================

def merge_coco_datasets(datasets_info, output_dir='/content/merged_10class_dataset'):
    """
    Merge multiple COCO datasets into one unified dataset
    """

    print("\n" + "=" * 70)
    print("🔄 MERGING ALL DATASETS INTO ONE")
    print("=" * 70)

    # Create output directories
    os.makedirs(output_dir, exist_ok=True)
    for split in ['train', 'valid', 'test']:
        os.makedirs(os.path.join(output_dir, split), exist_ok=True)

    # Merge each split
    for split in ['train', 'valid', 'test']:
        print(f"\n🔄 Processing {split.upper()} split...")

        merged_images = []
        merged_annotations = []

        image_id_counter = 1
        annotation_id_counter = 1

        split_stats = []

        for dataset_info in datasets_info:
            dataset_name = dataset_info['name']
            dataset_path = dataset_info['path']
            class_id = dataset_info['class_id']

            # Path to annotations
            split_path = os.path.join(dataset_path, split)
            json_path = os.path.join(split_path, '_annotations.coco.json')

            if not os.path.exists(json_path):
                print(f"   ⚠️  {dataset_name}/{split} not found, skipping...")
                continue

            # Load COCO JSON
            with open(json_path, 'r') as f:
                coco_data = json.load(f)

            # Map old image IDs to new ones
            old_to_new_image_id = {}

            # Process images
            for img in coco_data['images']:
                old_img_id = img['id']
                new_img_id = image_id_counter
                old_to_new_image_id[old_img_id] = new_img_id
                image_id_counter += 1

                # Copy image file with unique name
                old_img_path = os.path.join(split_path, img['file_name'])
                file_ext = os.path.splitext(img['file_name'])[1]
                new_img_name = f"{dataset_name}_{new_img_id}{file_ext}"
                new_img_path = os.path.join(output_dir, split, new_img_name)

                if os.path.exists(old_img_path):
                    shutil.copy(old_img_path, new_img_path)

                # Add to merged images
                merged_images.append({
                    'id': new_img_id,
                    'file_name': new_img_name,
                    'width': img.get('width', 640),
                    'height': img.get('height', 640)
                })

            # Process annotations
            ann_count = 0
            for ann in coco_data['annotations']:
                if ann['image_id'] not in old_to_new_image_id:
                    continue

                new_img_id = old_to_new_image_id[ann['image_id']]

                merged_annotations.append({
                    'id': annotation_id_counter,
                    'image_id': new_img_id,
                    'category_id': class_id,  # Our consistent class ID
                    'bbox': ann['bbox'],
                    'area': ann.get('area', ann['bbox'][2] * ann['bbox'][3]),
                    'iscrowd': ann.get('iscrowd', 0)
                })

                annotation_id_counter += 1
                ann_count += 1

            img_count = len(coco_data['images'])
            split_stats.append(f"{dataset_name}: {img_count} imgs, {ann_count} anns")
            print(f"   ✅ {dataset_name:15s} → {img_count:4d} images, {ann_count:4d} annotations")

        # Create merged COCO JSON
        merged_coco = {
            'images': merged_images,
            'annotations': merged_annotations,
            'categories': [
                {'id': ds['class_id'], 'name': ds['name'], 'supercategory': 'object'}
                for ds in datasets_info
            ]
        }

        # Save merged JSON
        output_json = os.path.join(output_dir, split, '_annotations.coco.json')
        with open(output_json, 'w') as f:
            json.dump(merged_coco, f, indent=2)

        print(f"\n   📊 {split.upper()} TOTAL: {len(merged_images)} images, {len(merged_annotations)} annotations")

    print("\n" + "=" * 70)
    print(f"✅ MERGE COMPLETE!")
    print(f"📁 Merged dataset saved to: {output_dir}")
    print("=" * 70)

    return output_dir

# Run the merge
merged_dataset_path = merge_coco_datasets(downloaded_datasets)
```

```python
#  STEP 4: ANALYZE MERGED DATASET
# ========================================

def analyze_merged_dataset(dataset_path):
    """Detailed analysis of the merged dataset"""

    print("\n" + "=" * 70)
    print("MERGED DATASET ANALYSIS")
    print("=" * 70)

    total_images = 0
    total_annotations = 0

    for split in ['train', 'valid', 'test']:
        json_path = os.path.join(dataset_path, split, '_annotations.coco.json')

        if not os.path.exists(json_path):
            continue

        with open(json_path, 'r') as f:
            coco = json.load(f)

        num_images = len(coco['images'])
        num_anns = len(coco['annotations'])
        total_images += num_images
        total_annotations += num_anns

        print(f"\n{'='*70}")
        print(f"{split.upper()} SPLIT:")
        print(f"{'='*70}")
        print(f"   Total Images: {num_images}")
        print(f"   Total Annotations: {num_anns}")
        print(f"   Avg Objects/Image: {num_anns/num_images:.2f}")

        # Objects per image distribution
        from collections import defaultdict
        img_ann_count = defaultdict(int)
        for ann in coco['annotations']:
            img_ann_count[ann['image_id']] += 1

        obj_counts = list(img_ann_count.values())
        if obj_counts:
            print(f" Objects per image: Min={min(obj_counts)}, Max={max(obj_counts)}, Avg={sum(obj_counts)/len(obj_counts):.2f}")

        # Class distribution
        class_counts = Counter([ann['category_id'] for ann in coco['annotations']])

        print(f"\n  CLASS DISTRIBUTION:")
        for cat in sorted(coco['categories'], key=lambda x: x['id']):
            count = class_counts.get(cat['id'], 0)
            percentage = (count / num_anns * 100) if num_anns > 0 else 0
            print(f"     [{cat['id']}] {cat['name']:15s}: {count:5d} objects ({percentage:5.2f}%)")

        # Bounding box statistics
        if coco['annotations']:
            widths = [ann['bbox'][2] for ann in coco['annotations']]
            heights = [ann['bbox'][3] for ann in coco['annotations']]
            areas = [ann['bbox'][2] * ann['bbox'][3] for ann in coco['annotations']]

            print(f"\n   BOUNDING BOX STATISTICS:")
            print(f"     Width  - Mean: {sum(widths)/len(widths):6.1f}, Min: {min(widths):6.1f}, Max: {max(widths):6.1f}")
            print(f"     Height - Mean: {sum(heights)/len(heights):6.1f}, Min: {min(heights):6.1f}, Max: {max(heights):6.1f}")
            print(f"     Area   - Mean: {sum(areas)/len(areas):8.1f}, Min: {min(areas):8.1f}, Max: {max(areas):8.1f}")

    print("\n" + "=" * 70)
    print(f"   OVERALL SUMMARY:")
    print(f"   Total Images: {total_images}")
    print(f"   Total Annotations: {total_annotations}")
    print(f"   Classes: 10")
    print("=" * 70)

# Run analysis
analyze_merged_dataset(merged_dataset_path)
```

```python
# =======================================
# IMPORTS & SETUP


import os
import json
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import cv2
import matplotlib.pyplot as plt
from glob import glob
import albumentations as A

print("TensorFlow version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))

# Set mixed precision for faster training
from tensorflow.keras import mixed_precision
policy = mixed_precision.Policy('mixed_float16')
mixed_precision.set_global_policy(policy)
print('Compute dtype: %s' % policy.compute_dtype)
print('Variable dtype: %s' % policy.variable_dtype)

# CONFIGURATION


class Config:
    # Paths
    DATASET_PATH = '/content/drive/MyDrive/FYP_10Class_Dataset'  # Your merged dataset

    # Model parameters
    INPUT_SIZE = 640
    GRID_SIZE = 20  # 20x20 grid for predictions
    NUM_CLASSES = 10
    NUM_ANCHORS = 3  # 3 anchor boxes per grid cell

    # Training parameters
    BATCH_SIZE = 8  # Adjust based on GPU memory
    EPOCHS = 150
    LEARNING_RATE = 1e-3

    # Class names (in order 0-9)
    CLASS_NAMES = [
        'cans', 'fire', 'traffic_light', 'pothole', 'helmet',
        'pen', 'e_waste', 'bags', 'water_bottle', 'knife'
    ]

    # Class weights (to handle imbalance)
    # Calculated as: max_count / class_count, normalized
    CLASS_WEIGHTS = {
        0: 2.12,   # cans
        1: 1.0,    # fire (most samples, baseline)
        2: 3.90,   # traffic_light
        3: 2.19,   # pothole
        4: 2.15,   # helmet
        5: 1.93,   # pen
        6: 4.99,   # e_waste
        7: 11.14,  # bags (least samples)
        8: 3.98,   # water_bottle
        9: 5.91,   # knife
    }

    # Anchor boxes (will calculate from data later)
    # Format: (width, height) in pixels
    ANCHORS = [
        # Small objects
        (30, 40), (50, 60), (40, 80),
        # Medium objects
        (100, 120), (150, 180), (120, 200),
        # Large objects
        (250, 300), (350, 400), (500, 550)
    ]

cfg = Config()

# DATA AUGMENTATION PIPELINE


# Training augmentation (aggressive)
# train_transform = A.Compose([
#     # Geometric
#     A.HorizontalFlip(p=0.5),
#     A.ShiftScaleRotate(
#         shift_limit=0.1,
#         scale_limit=0.2,
#         rotate_limit=15,
#         border_mode=cv2.BORDER_CONSTANT,
#         value=0,
#         p=0.5
#     ),

#     # Color
#     A.RandomBrightnessContrast(
#         brightness_limit=0.3,
#         contrast_limit=0.3,
#         p=0.5
#     ),
#     A.HueSaturationValue(
#         hue_shift_limit=20,
#         sat_shift_limit=30,
#         val_shift_limit=20,
#         p=0.4
#     ),

#     # Blur & Noise
#     A.OneOf([
#         A.MotionBlur(blur_limit=5, p=1.0),
#         A.GaussianBlur(blur_limit=5, p=1.0),
#         A.MedianBlur(blur_limit=5, p=1.0),
#     ], p=0.3),

#     A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),

#     # Cutout
#     A.CoarseDropout(
#         max_holes=3,
#         max_height=64,
#         max_width=64,
#         min_holes=1,
#         min_height=32,
#         min_width=32,
#         fill_value=0,
#         p=0.3
#     ),

# ], bbox_params=A.BboxParams(
#     format='coco',
#     label_fields=['class_labels'],
#     min_visibility=0.3  # Keep boxes with 30%+ visible
# ))

# DATA AUGMENTATION PIPELINE (FIXED)

# Training augmentation (with bbox clipping)
train_transform = A.Compose([
    # Geometric (REDUCED to prevent out-of-bounds)
    A.HorizontalFlip(p=0.5),
    A.ShiftScaleRotate(
        shift_limit=0.05,      # ← CHANGED from 0.1
        scale_limit=0.15,      # ← CHANGED from 0.2
        rotate_limit=10,       # ← CHANGED from 15
        border_mode=cv2.BORDER_CONSTANT,
        value=0,
        p=0.5
    ),

    # Color
    A.RandomBrightnessContrast(
        brightness_limit=0.3,
        contrast_limit=0.3,
        p=0.5
    ),
    A.HueSaturationValue(
        hue_shift_limit=20,
        sat_shift_limit=30,
        val_shift_limit=20,
        p=0.4
    ),

    # Blur & Noise
    A.OneOf([
        A.MotionBlur(blur_limit=5, p=1.0),
        A.GaussianBlur(blur_limit=5, p=1.0),
        A.MedianBlur(blur_limit=5, p=1.0),
    ], p=0.3),

    A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),

    # Cutout
    A.CoarseDropout(
        max_holes=3,
        max_height=64,
        max_width=64,
        min_holes=1,
        min_height=32,
        min_width=32,
        fill_value=0,
        p=0.3
    ),

], bbox_params=A.BboxParams(
    format='coco',
    label_fields=['class_labels'],
    min_visibility=0.3,
    clip=True  # ← ADDED: Enable bbox clipping
))

# Validation augmentation (none - just normalization)
val_transform = A.Compose([
    # No augmentation for validation
], bbox_params=A.BboxParams(
    format='coco',
    label_fields=['class_labels'],
    clip=True  # ← ADDED: Enable clipping
))

# Validation augmentation (none - just normalization)
val_transform = A.Compose([
    # No augmentation for validation
], bbox_params=A.BboxParams(
    format='coco',
    label_fields=['class_labels']
))

# DATA LOADING & PREPROCESSING

def load_coco_annotations(json_path, img_dir):
    """Load COCO format annotations"""
    with open(json_path, 'r') as f:
        coco_data = json.load(f)

    # Create image_id to filename mapping
    id_to_file = {img['id']: img['file_name'] for img in coco_data['images']}

    # Group annotations by image
    annotations_by_image = {}
    for ann in coco_data['annotations']:
        img_id = ann['image_id']
        if img_id not in annotations_by_image:
            annotations_by_image[img_id] = []
        annotations_by_image[img_id].append(ann)

    # Create dataset records
    records = []
    for img_id, anns in annotations_by_image.items():
        if img_id not in id_to_file:
            continue

        img_path = os.path.join(img_dir, id_to_file[img_id])

        if not os.path.exists(img_path):
            continue

        boxes = []
        classes = []

        for ann in anns:
            x, y, w, h = ann['bbox']
            # Ensure valid boxes
            if w > 0 and h > 0:
                boxes.append([float(x), float(y), float(w), float(h)])
                classes.append(int(ann['category_id']))

        if len(boxes) > 0:
            records.append({
                'image_path': img_path,
                'boxes': np.array(boxes, dtype=np.float32),  #to numpy array
                'classes': np.array(classes, dtype=np.int32)
            })

    return records

def encode_to_grid(boxes, classes, grid_size=20, input_size=640, num_classes=10):
    """
    Encode bounding boxes to grid format for training

    Args:
        boxes: numpy array or list of [x, y, w, h]
        classes: numpy array or list of class IDs

    Returns:
        grid: [grid_size, grid_size, num_anchors, (5 + num_classes)]
    """
    num_anchors = 3  # Using 3 anchors
    grid = np.zeros((grid_size, grid_size, num_anchors, 5 + num_classes), dtype=np.float32)

    # Convert to numpy arrays if needed
    if not isinstance(boxes, np.ndarray):
        boxes = np.array(boxes, dtype=np.float32)
    if not isinstance(classes, np.ndarray):
        classes = np.array(classes, dtype=np.int32)

    # Handle empty boxes
    if len(boxes) == 0:
        return grid

    cell_size = input_size / grid_size

    # Use first 3 anchors (small/medium scale)
    anchors = np.array(cfg.ANCHORS[:3], dtype=np.float32)

    for box, cls in zip(boxes, classes):
        # Ensure box is array-like
        if isinstance(box, (tuple, list)):
            box = np.array(box, dtype=np.float32)

        x, y, w, h = float(box[0]), float(box[1]), float(box[2]), float(box[3])

        # Skip invalid boxes
        if w <= 0 or h <= 0:
            continue

        # Convert to center coordinates
        cx = x + w / 2
        cy = y + h / 2

        # Find which grid cell this object belongs to
        grid_x = int(np.clip(cx / cell_size, 0, grid_size - 1))
        grid_y = int(np.clip(cy / cell_size, 0, grid_size - 1))

        # Find best matching anchor
        best_anchor = 0
        best_iou = 0

        for anchor_idx in range(num_anchors):
            anchor_w, anchor_h = anchors[anchor_idx]

            # Calculate IoU (intersection over union)
            intersect_w = min(w, anchor_w)
            intersect_h = min(h, anchor_h)
            intersect = intersect_w * intersect_h

            union = w * h + anchor_w * anchor_h - intersect
            iou = intersect / (union + 1e-6)

            if iou > best_iou:
                best_iou = iou
                best_anchor = anchor_idx

        # Encode box relative to grid cell
        x_cell = (cx % cell_size) / cell_size  # 0-1 within cell
        y_cell = (cy % cell_size) / cell_size
        w_anchor = np.log((w + 1e-6) / (anchors[best_anchor][0] + 1e-6))
        h_anchor = np.log((h + 1e-6) / (anchors[best_anchor][1] + 1e-6))

        # Ensure class is valid
        cls = int(cls)
        if cls < 0 or cls >= num_classes:
            continue

        # Encode to grid (only if cell not already occupied)
        if grid[grid_y, grid_x, best_anchor, 4] == 0:  # Cell not occupied
            grid[grid_y, grid_x, best_anchor, 0] = x_cell
            grid[grid_y, grid_x, best_anchor, 1] = y_cell
            grid[grid_y, grid_x, best_anchor, 2] = w_anchor
            grid[grid_y, grid_x, best_anchor, 3] = h_anchor
            grid[grid_y, grid_x, best_anchor, 4] = 1.0  # Objectness
            grid[grid_y, grid_x, best_anchor, 5 + cls] = 1.0  # Class one-hot

    return grid

def preprocess_image_and_boxes(record, transform, training=True):
    """Load image, apply augmentation, encode to grid"""

    try:
        # Load image
        image = cv2.imread(record['image_path'])
        if image is None:
            raise ValueError(f"Failed to load image: {record['image_path']}")

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Get boxes and classes as lists (for albumentations)
        boxes = record['boxes'].tolist() if isinstance(record['boxes'], np.ndarray) else record['boxes']
        classes = record['classes'].tolist() if isinstance(record['classes'], np.ndarray) else record['classes']

        # Apply augmentation
        if training and transform is not None:
            try:
                transformed = transform(
                    image=image,
                    bboxes=boxes,
                    class_labels=classes
                )
                image = transformed['image']
                boxes = transformed['bboxes']
                classes = transformed['class_labels']
            except Exception as e:
                # If augmentation fails, use original
                print(f"Augmentation warning: {e}")

        # Convert boxes back to numpy array
        if len(boxes) > 0:
            boxes = np.array(boxes, dtype=np.float32)
            classes = np.array(classes, dtype=np.int32)
        else:
            boxes = np.zeros((0, 4), dtype=np.float32)
            classes = np.array([], dtype=np.int32)

        # Resize image to input size
        h_orig, w_orig = image.shape[:2]
        image = cv2.resize(image, (cfg.INPUT_SIZE, cfg.INPUT_SIZE))

        # Scale boxes to match resized image
        if len(boxes) > 0:
            # Note: boxes are already in resized coordinates after augmentation
            # But we need to ensure they're within bounds
            boxes[:, 0] = np.clip(boxes[:, 0], 0, cfg.INPUT_SIZE - 1)  # x
            boxes[:, 1] = np.clip(boxes[:, 1], 0, cfg.INPUT_SIZE - 1)  # y
            boxes[:, 2] = np.clip(boxes[:, 2], 1, cfg.INPUT_SIZE)      # w
            boxes[:, 3] = np.clip(boxes[:, 3], 1, cfg.INPUT_SIZE)      # h

        # Normalize image
        image = image.astype(np.float32) / 255.0

        # Encode boxes to grid
        grid = encode_to_grid(boxes, classes, cfg.GRID_SIZE, cfg.INPUT_SIZE, cfg.NUM_CLASSES)

        return image, grid

    except Exception as e:
        # Return empty data on error
        print(f"Error in preprocess: {e}")
        image = np.zeros((cfg.INPUT_SIZE, cfg.INPUT_SIZE, 3), dtype=np.float32)
        grid = np.zeros((cfg.GRID_SIZE, cfg.GRID_SIZE, 3, 5 + cfg.NUM_CLASSES), dtype=np.float32)
        return image, grid

# ========================================
# DATASET CREATION (FIXED)
# ========================================

def create_dataset(split='train', batch_size=8, shuffle=True):
    """Create TensorFlow dataset"""

    # Load annotations
    json_path = os.path.join(cfg.DATASET_PATH, split, '_annotations.coco.json')
    img_dir = os.path.join(cfg.DATASET_PATH, split)

    print(f"Loading {split} annotations from: {json_path}")
    records = load_coco_annotations(json_path, img_dir)
    print(f"Loaded {len(records)} images for {split}")

    # Choose augmentation
    transform = train_transform if split == 'train' else val_transform
    training = (split == 'train')

    def generator():
        indices = list(range(len(records)))
        if shuffle:
            np.random.shuffle(indices)

        success_count = 0
        error_count = 0

        for idx in indices:
            record = records[idx]
            try:
                image, grid = preprocess_image_and_boxes(record, transform, training)

                # Verify valid data
                if np.isnan(image).any() or np.isnan(grid).any():
                    error_count += 1
                    continue

                success_count += 1
                yield image, grid

            except Exception as e:
                error_count += 1
                if error_count <= 5:  # Only print first 5 errors
                    print(f"Error processing {record['image_path']}: {e}")
                continue

        print(f"{split} generator: {success_count} successful, {error_count} errors")

    # Create TF dataset
    output_signature = (
        tf.TensorSpec(shape=(cfg.INPUT_SIZE, cfg.INPUT_SIZE, 3), dtype=tf.float32),
        tf.TensorSpec(shape=(cfg.GRID_SIZE, cfg.GRID_SIZE, 3, 5 + cfg.NUM_CLASSES), dtype=tf.float32)
    )

    dataset = tf.data.Dataset.from_generator(
        generator,
        output_signature=output_signature
    )

    dataset = dataset.batch(batch_size, drop_remainder=False)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset, len(records)

# ========================================
# LOAD DATASETS (FIXED)
# ========================================

print("\n" + "="*70)
print(" CREATING DATASETS")
print("="*70)

train_dataset, train_size = create_dataset('train', batch_size=cfg.BATCH_SIZE, shuffle=True)
valid_dataset, valid_size = create_dataset('valid', batch_size=cfg.BATCH_SIZE, shuffle=False)

print(f"\n✅ Datasets created successfully!")
print(f"   Train: {train_size} images")
print(f"   Valid: {valid_size} images")

# Visualize a batch to verify
print("\n🔍 Verifying data pipeline...")
for images, grids in train_dataset.take(1):
    print(f"✅ Image batch shape: {images.shape}")
    print(f"✅ Grid batch shape: {grids.shape}")
    print(f"✅ Image value range: [{images.numpy().min():.3f}, {images.numpy().max():.3f}]")

    # Count cells with objects
    obj_mask = grids.numpy()[..., 4] > 0
    num_obj_cells = obj_mask.sum()
    print(f"✅ Grid contains {num_obj_cells} cells with objects")

    # Show per-image object counts
    for i in range(min(3, images.shape[0])):
        obj_count = obj_mask[i].sum()
        print(f"   Image {i}: {obj_count} object cells")

print("\n" + "="*70)
print("✅ DATA PIPELINE READY!")
print("="*70)
```

```python
# ========================================
# MODEL ARCHITECTURE - CUSTOM DETECTOR
# ========================================

from tensorflow.keras import layers, Model
import tensorflow as tf

def conv_block(x, filters, kernel_size=3, strides=1, activation='leaky_relu', batch_norm=True, name=''):
    """Convolutional block with BatchNorm and activation"""

    x = layers.Conv2D(
        filters,
        kernel_size,
        strides=strides,
        padding='same',
        use_bias=not batch_norm,
        kernel_initializer='he_normal',
        name=f'{name}_conv' if name else None
    )(x)

    if batch_norm:
        x = layers.BatchNormalization(name=f'{name}_bn' if name else None)(x)

    if activation == 'leaky_relu':
        x = layers.LeakyReLU(alpha=0.1, name=f'{name}_act' if name else None)(x)
    elif activation == 'relu':
        x = layers.ReLU(name=f'{name}_act' if name else None)(x)
    elif activation == 'swish':
        x = layers.Activation('swish', name=f'{name}_act' if name else None)(x)

    return x

def residual_block(x, filters, name=''):
    """Residual block with skip connection"""
    shortcut = x

    # First conv (reduce channels)
    x = conv_block(x, filters // 2, kernel_size=1, name=f'{name}_conv1')

    # Second conv (restore channels)
    x = conv_block(x, filters, kernel_size=3, name=f'{name}_conv2')

    # Skip connection
    x = layers.Add(name=f'{name}_add')([shortcut, x])

    return x

def build_detector_model(input_shape=(640, 640, 3), num_classes=10, num_anchors=3):
    """
    Build custom object detection model from scratch

    Architecture:
    - Backbone: Custom CNN with residual connections
    - Neck: Feature pyramid-like structure
    - Head: Detection head for each grid cell

    Output: [batch, 20, 20, 3, 15]
            15 = [x, y, w, h, objectness, class_0...class_9]
    """

    inputs = layers.Input(shape=input_shape, name='input')

    print("\n" + "="*70)
    print("🏗️  BUILDING MODEL ARCHITECTURE")
    print("="*70)


    # BACKBONE (Feature Extraction)


    print("\n Building Backbone...")

    # Stem: 640×640×3 → 640×640×32
    x = conv_block(inputs, 32, kernel_size=3, strides=1, name='stem')
    print(f"   Stem: {x.shape}")

    # Stage 1: 640×640×32 → 320×320×64
    x = conv_block(x, 64, kernel_size=3, strides=2, name='stage1_downsample')
    x = residual_block(x, 64, name='stage1_res1')
    print(f"   Stage 1: {x.shape}")

    # Stage 2: 320×320×64 → 160×160×128
    x = conv_block(x, 128, kernel_size=3, strides=2, name='stage2_downsample')
    x = residual_block(x, 128, name='stage2_res1')
    x = residual_block(x, 128, name='stage2_res2')
    print(f"   Stage 2: {x.shape}")

    # Stage 3: 160×160×128 → 80×80×256
    x = conv_block(x, 256, kernel_size=3, strides=2, name='stage3_downsample')
    x = residual_block(x, 256, name='stage3_res1')
    x = residual_block(x, 256, name='stage3_res2')
    x = residual_block(x, 256, name='stage3_res3')
    feature_80 = x  # Save for potential multi-scale
    print(f"   Stage 3: {x.shape}")

    # Stage 4: 80×80×256 → 40×40×512
    x = conv_block(x, 512, kernel_size=3, strides=2, name='stage4_downsample')
    x = residual_block(x, 512, name='stage4_res1')
    x = residual_block(x, 512, name='stage4_res2')
    x = residual_block(x, 512, name='stage4_res3')
    feature_40 = x  # Save for potential multi-scale
    print(f"   Stage 4: {x.shape}")

    # Stage 5: 40×40×512 → 20×20×512
    x = conv_block(x, 512, kernel_size=3, strides=2, name='stage5_downsample')
    x = residual_block(x, 512, name='stage5_res1')
    x = residual_block(x, 512, name='stage5_res2')
    feature_20 = x  # Main detection feature map
    print(f"   Stage 5 (Detection): {x.shape}")

    # ========================================
    # NECK (Feature Refinement)
    # ========================================

    print("\n Building Neck...")

    # Additional conv layers to refine features
    x = conv_block(feature_20, 512, kernel_size=3, name='neck_conv1')
    x = conv_block(x, 512, kernel_size=1, name='neck_conv2')
    x = conv_block(x, 512, kernel_size=3, name='neck_conv3')
    print(f"   Neck: {x.shape}")


    # DETECTION HEAD


    print("\n Building Detection Head...")

    # Detection head
    # Output: [batch, 20, 20, num_anchors * (5 + num_classes)]
    num_outputs = num_anchors * (5 + num_classes)  # 3 * 15 = 45

    # Final conv layers
    x = conv_block(x, 256, kernel_size=1, name='head_conv1')
    x = conv_block(x, 512, kernel_size=3, name='head_conv2')

    # Output layer (no activation - will apply in loss function)
    output = layers.Conv2D(
        num_outputs,
        kernel_size=1,
        padding='same',
        kernel_initializer='glorot_uniform',
        name='output'
    )(x)

    # Reshape to [batch, 20, 20, 3, 15]
    output = layers.Reshape((20, 20, num_anchors, 5 + num_classes), name='reshape')(output)

    print(f"   Output: {output.shape}")

    # ========================================
    # CREATE MODEL
    # ========================================

    model = Model(inputs=inputs, outputs=output, name='ObjectDetector')

    print("\n" + "="*70)
    print("✅ MODEL BUILT SUCCESSFULLY!")
    print("="*70)

    return model


# BUILD THE MODEL


print("\n" + "="*70)
print("INITIALIZING MODEL")
print("="*70)

model = build_detector_model(
    input_shape=(cfg.INPUT_SIZE, cfg.INPUT_SIZE, 3),
    num_classes=cfg.NUM_CLASSES,
    num_anchors=3
)

# Model summary
print("\n MODEL SUMMARY:")
model.summary()

# Count parameters
total_params = model.count_params()
print(f"\nTotal Parameters: {total_params:,}")
print(f"   Trainable: {sum([tf.keras.backend.count_params(w) for w in model.trainable_weights]):,}")
print(f"   Non-trainable: {sum([tf.keras.backend.count_params(w) for w in model.non_trainable_weights]):,}")

# Estimate model size
model_size_mb = (total_params * 4) / (1024 * 1024)  # 4 bytes per float32
print(f"\n Estimated Model Size: {model_size_mb:.2f} MB")

# Test forward pass
print("\nTesting forward pass...")
dummy_input = tf.random.normal((1, 640, 640, 3))
dummy_output = model(dummy_input, training=False)
print(f"✅ Forward pass successful!")
print(f"   Input shape: {dummy_input.shape}")
print(f"   Output shape: {dummy_output.shape}")

print("\n" + "="*70)
print("✅ MODEL READY FOR TRAINING!")
print("="*70)
```

```python
# ========================================
# LOSS FUNCTIONS - FOCAL LOSS + GIoU
# ========================================

import tensorflow as tf
from tensorflow import keras

def compute_iou(box1, box2):
    """
    Compute IoU between two boxes

    Args:
        box1, box2: [x, y, w, h] format

    Returns:
        iou: Intersection over Union
    """
    # Convert to [x1, y1, x2, y2]
    box1_x1 = box1[..., 0] - box1[..., 2] / 2
    box1_y1 = box1[..., 1] - box1[..., 3] / 2
    box1_x2 = box1[..., 0] + box1[..., 2] / 2
    box1_y2 = box1[..., 1] + box1[..., 3] / 2

    box2_x1 = box2[..., 0] - box2[..., 2] / 2
    box2_y1 = box2[..., 1] - box2[..., 3] / 2
    box2_x2 = box2[..., 0] + box2[..., 2] / 2
    box2_y2 = box2[..., 1] + box2[..., 3] / 2

    # Intersection area
    inter_x1 = tf.maximum(box1_x1, box2_x1)
    inter_y1 = tf.maximum(box1_y1, box2_y1)
    inter_x2 = tf.minimum(box1_x2, box2_x2)
    inter_y2 = tf.minimum(box1_y2, box2_y2)

    inter_w = tf.maximum(0.0, inter_x2 - inter_x1)
    inter_h = tf.maximum(0.0, inter_y2 - inter_y1)
    inter_area = inter_w * inter_h

    # Union area
    box1_area = box1[..., 2] * box1[..., 3]
    box2_area = box2[..., 2] * box2[..., 3]
    union_area = box1_area + box2_area - inter_area

    # IoU
    iou = inter_area / (union_area + 1e-7)

    return iou

def compute_giou(box1, box2):
    """
    Compute Generalized IoU (GIoU)

    GIoU is better than IoU because it considers:
    - The overlap (like IoU)
    - The shape and location of non-overlapping areas

    Range: [-1, 1] where 1 is perfect overlap
    """
    # Regular IoU
    iou = compute_iou(box1, box2)

    # Convert to [x1, y1, x2, y2]
    box1_x1 = box1[..., 0] - box1[..., 2] / 2
    box1_y1 = box1[..., 1] - box1[..., 3] / 2
    box1_x2 = box1[..., 0] + box1[..., 2] / 2
    box1_y2 = box1[..., 1] + box1[..., 3] / 2

    box2_x1 = box2[..., 0] - box2[..., 2] / 2
    box2_y1 = box2[..., 1] - box2[..., 3] / 2
    box2_x2 = box2[..., 0] + box2[..., 2] / 2
    box2_y2 = box2[..., 1] + box2[..., 3] / 2

    # Smallest enclosing box
    enclose_x1 = tf.minimum(box1_x1, box2_x1)
    enclose_y1 = tf.minimum(box1_y1, box2_y1)
    enclose_x2 = tf.maximum(box1_x2, box2_x2)
    enclose_y2 = tf.maximum(box1_y2, box2_y2)

    enclose_w = enclose_x2 - enclose_x1
    enclose_h = enclose_y2 - enclose_y1
    enclose_area = enclose_w * enclose_h

    # GIoU
    box1_area = box1[..., 2] * box1[..., 3]
    box2_area = box2[..., 2] * box2[..., 3]
    union_area = box1_area + box2_area - (iou * (box1_area + box2_area - iou * enclose_area))

    giou = iou - (enclose_area - union_area) / (enclose_area + 1e-7)

    return giou

def focal_loss(y_true, y_pred, alpha=0.25, gamma=2.0):
    """
    Focal Loss for classification

    Focuses training on hard examples and down-weights easy examples.
    Essential for handling class imbalance!

    Formula: FL = -alpha * (1-pt)^gamma * log(pt)

    Args:
        y_true: Ground truth labels [0 or 1]
        y_pred: Predicted probabilities [0-1]
        alpha: Balancing factor (0.25 works well)
        gamma: Focusing parameter (2.0 standard)

    Returns:
        loss: Focal loss value
    """
    # Clip predictions to prevent log(0)
    y_pred = tf.clip_by_value(y_pred, 1e-7, 1 - 1e-7)

    # Binary cross entropy
    bce = - (y_true * tf.math.log(y_pred) + (1 - y_true) * tf.math.log(1 - y_pred))

    # Focal term
    pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
    focal_weight = alpha * tf.pow(1 - pt, gamma)

    # Focal loss
    loss = focal_weight * bce

    return loss

class DetectionLoss(keras.losses.Loss):
    """
    Complete detection loss combining:
    1. Localization Loss (GIoU)
    2. Objectness Loss (Focal Loss)
    3. Classification Loss (Focal Loss)
    """

    def __init__(self,
                 num_classes=10,
                 lambda_coord=5.0,
                 lambda_obj=1.0,
                 lambda_class=1.0,
                 lambda_noobj=0.5,
                 class_weights=None,
                 name='detection_loss'):

        super().__init__(name=name)

        self.num_classes = num_classes
        self.lambda_coord = lambda_coord
        self.lambda_obj = lambda_obj
        self.lambda_class = lambda_class
        self.lambda_noobj = lambda_noobj

        # Track metrics MANUALLY (required by TF 2.12+)
        self.giou_metric = keras.metrics.Mean(name="giou_loss")
        self.obj_metric = keras.metrics.Mean(name="obj_loss")
        self.noobj_metric = keras.metrics.Mean(name="noobj_loss")
        self.class_metric = keras.metrics.Mean(name="class_loss")

        # Class weights
        if class_weights is None:
            self.class_weights = tf.ones(num_classes, dtype=tf.float32)
        else:
            self.class_weights = tf.constant(list(class_weights.values()), dtype=tf.float32)

    @property
    def metrics(self):
        """Tell Keras which metrics to log."""
        return [
            self.giou_metric,
            self.obj_metric,
            self.noobj_metric,
            self.class_metric
        ]

    def call(self, y_true, y_pred):

        # -----------------------------
        # Split predictions
        # -----------------------------
        pred_xy = tf.sigmoid(y_pred[..., 0:2])
        pred_wh = y_pred[..., 2:4]
        pred_obj = tf.sigmoid(y_pred[..., 4:5])
        pred_class = tf.sigmoid(y_pred[..., 5:])

        true_xy = y_true[..., 0:2]
        true_wh = y_true[..., 2:4]
        true_obj = y_true[..., 4:5]
        true_class = y_true[..., 5:]

        obj_mask = true_obj
        noobj_mask = 1.0 - obj_mask

        # -----------------------------
        # 1. GIoU Loss
        # -----------------------------
        pred_boxes = tf.concat([pred_xy, tf.exp(pred_wh)], axis=-1)
        true_boxes = tf.concat([true_xy, tf.exp(true_wh)], axis=-1)

        giou = compute_giou(pred_boxes, true_boxes)[..., tf.newaxis]

        giou_loss = obj_mask * (1.0 - giou)
        giou_loss = tf.reduce_sum(giou_loss) / (tf.reduce_sum(obj_mask) + 1e-7)

        self.giou_metric.update_state(giou_loss)

        # -----------------------------
        # 2. Objectness (Focal Loss)
        # -----------------------------
        obj_loss = obj_mask * focal_loss(true_obj, pred_obj)
        obj_loss = tf.reduce_sum(obj_loss) / (tf.reduce_sum(obj_mask) + 1e-7)
        self.obj_metric.update_state(obj_loss)

        # No-objectness
        noobj_loss = noobj_mask * focal_loss(true_obj, pred_obj)
        noobj_loss = tf.reduce_sum(noobj_loss) / (tf.reduce_sum(noobj_mask) + 1e-7)
        self.noobj_metric.update_state(noobj_loss)

        # -----------------------------
        # 3. Classification Loss
        # -----------------------------
        class_loss = focal_loss(true_class, pred_class)

        cw = tf.reshape(self.class_weights, [1, 1, 1, 1, self.num_classes])
        weighted_class_loss = class_loss * cw * true_class

        weighted_class_loss = tf.reduce_sum(weighted_class_loss, axis=-1, keepdims=True)
        weighted_class_loss = obj_mask * weighted_class_loss

        class_loss_final = tf.reduce_sum(weighted_class_loss) / (tf.reduce_sum(obj_mask) + 1e-7)
        self.class_metric.update_state(class_loss_final)

        # -----------------------------
        # FINAL LOSS
        # -----------------------------
        total_loss = (
            self.lambda_coord * giou_loss +
            self.lambda_obj * obj_loss +
            self.lambda_noobj * noobj_loss +
            self.lambda_class * class_loss_final
        )

        return total_loss


# RE-INSTANTIATE LOSS FUNCTION


print("\n" + "="*70)
print("CREATING FIXED LOSS FUNCTION")
print("="*70)

loss_fn = DetectionLoss(
    num_classes=cfg.NUM_CLASSES,
    lambda_coord=5.0,
    lambda_obj=1.0,
    lambda_class=1.0,
    lambda_noobj=0.5,
    class_weights=cfg.CLASS_WEIGHTS
)

print("\n✅ Loss Function Ready!")

# Test loss function again
print("\n Testing loss function...")
for images, grids in train_dataset.take(1):
    # Forward pass
    predictions = model(images, training=True)

    # Compute loss
    loss_value = loss_fn(grids, predictions)

    print(f"✅ Loss computation successful!")
    print(f"   Batch size: {images.shape[0]}")
    print(f"   Total Loss: {loss_value.numpy():.4f}")

    # Check individual losses
    print(f"\n   Individual Losses:")
    for metric in loss_fn.metrics:
        print(f"      {metric.name}: {metric.result().numpy():.4f}")

print("\n" + "="*70)
print("✅ LOSS FUNCTION VERIFIED!")
print("="*70)
```

```python
# ========================================
# TRAINING SETUP (FIXED - LOWER LR + GRADIENT CLIPPING)
# ========================================

import tensorflow as tf
from tensorflow import keras
import numpy as np
import time
from datetime import datetime
import os

print("\n" + "="*70)
print(" TRAINING SETUP (FIXED)")
print("="*70)

# ========================================
# GPU CHECK
# ========================================

print("\n🔍 Checking GPU...")
gpus = tf.config.list_physical_devices('GPU')
if len(gpus) > 0:
    print(f" GPU Available: {gpus[0].name}")
else:
    print("⚠️ WARNING: No GPU detected!")

# ========================================
# VERIFY DATASETS
# ========================================

print("\n🔍 Verifying datasets...")
try:
    for batch in train_dataset.take(1):
        print(f"✅ Train dataset: {batch[0].shape}")
    for batch in valid_dataset.take(1):
        print(f"✅ Valid dataset: {batch[0].shape}")
except Exception as e:
    print(f"⚠️ Dataset issue, recreating: {e}")
    train_dataset, _ = create_dataset('train', batch_size=cfg.BATCH_SIZE, shuffle=True)
    valid_dataset, _ = create_dataset('valid', batch_size=cfg.BATCH_SIZE, shuffle=False)
    print("✅ Datasets recreated!")

# ========================================
# LEARNING RATE SCHEDULE (FIXED - LOWER INITIAL LR)
# ========================================

def create_lr_schedule(initial_lr=5e-4, warmup_epochs=5, total_epochs=150):  # ← CHANGED from 1e-3 to 5e-4
    def lr_schedule(epoch):
        if epoch < warmup_epochs:
            return initial_lr * (epoch + 1) / warmup_epochs
        progress = (epoch - warmup_epochs) / (total_epochs - warmup_epochs)
        cosine_decay = 0.5 * (1 + np.cos(np.pi * progress))
        return 1e-6 + (initial_lr - 1e-6) * cosine_decay
    return lr_schedule

# Use the new lower learning rate
FIXED_LR = 5e-4  # ← NEW SAFER LEARNING RATE

lr_schedule_fn = create_lr_schedule(
    initial_lr=FIXED_LR,  # ← Using fixed LR instead of cfg.LEARNING_RATE
    warmup_epochs=5,
    total_epochs=cfg.EPOCHS
)

lr_callback = keras.callbacks.LearningRateScheduler(lr_schedule_fn, verbose=0)

print(f"\n✅ Learning Rate: {FIXED_LR} (lowered from 0.001 to prevent NaN)")

# ========================================
# CALLBACKS
# ========================================

checkpoint_path = "/content/drive/MyDrive/FYP_Model_Checkpoints/best_model.h5"
os.makedirs(os.path.dirname(checkpoint_path), exist_ok=True)

checkpoint_callback = keras.callbacks.ModelCheckpoint(
    filepath=checkpoint_path,
    monitor='val_loss',
    save_best_only=True,
    save_weights_only=False,
    mode='min',
    verbose=1
)

early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=20,
    restore_best_weights=True,
    verbose=1
)

log_dir = "/content/drive/MyDrive/FYP_Logs/" + datetime.now().strftime("%Y%m%d-%H%M%S")
os.makedirs(log_dir, exist_ok=True)

tensorboard_callback = keras.callbacks.TensorBoard(
    log_dir=log_dir,
    histogram_freq=1,
    update_freq='epoch'
)

csv_logger = keras.callbacks.CSVLogger(
    "/content/drive/MyDrive/FYP_Logs/training_history.csv",
    append=True
)

# ========================================
# NaN DETECTION CALLBACK (NEW!)
# ========================================

class NaNDetector(keras.callbacks.Callback):
    """Stop training if NaN is detected"""
    def on_batch_end(self, batch, logs=None):
        logs = logs or {}
        loss = logs.get('loss', 0)

        if np.isnan(loss) or np.isinf(loss):
            print(f"\n NaN/Inf detected at batch {batch}! Stopping training.")
            print(f"   Loss value: {loss}")
            print(f"   Try lowering learning rate further (use 2e-4 or 1e-4)")
            self.model.stop_training = True

nan_detector = NaNDetector()

# ========================================
# CLEAN EPOCH LOGGER
# ========================================

class CleanEpochLogger(keras.callbacks.Callback):
    def on_epoch_begin(self, epoch, logs=None):
        self.start_time = time.time()
        print(f"\n{'='*70}")
        print(f"EPOCH {epoch+1}/{cfg.EPOCHS}")
        print(f"{'='*70}")

    def on_epoch_end(self, epoch, logs=None):
        duration = time.time() - self.start_time
        logs = logs or {}

        print(f"\n{'='*70}")
        print(f"✅ EPOCH {epoch+1} COMPLETE - {duration:.1f}s")
        print(f"{'='*70}")

        # Check for NaN in metrics
        loss = logs.get('loss', 0)
        val_loss = logs.get('val_loss', 0)

        if np.isnan(loss) or np.isnan(val_loss):
            print("\n WARNING: NaN detected in losses!")
            print("   Training will stop. Try lowering LR to 2e-4 or 1e-4")
        else:
            print(f"""
📊 Training Metrics:
   Loss:      {logs.get('loss', 0):.4f}
   GIoU:      {logs.get('giou_loss', 0):.4f}
   Obj:       {logs.get('obj_loss', 0):.4f}
   NoObj:     {logs.get('noobj_loss', 0):.4f}
   Class:     {logs.get('class_loss', 0):.4f}

📊 Validation Metrics:
   Val Loss:  {logs.get('val_loss', 0):.4f}
   Val GIoU:  {logs.get('val_giou_loss', 0):.4f}
   Val Obj:   {logs.get('val_obj_loss', 0):.4f}
   Val NoObj: {logs.get('val_noobj_loss', 0):.4f}
   Val Class: {logs.get('val_class_loss', 0):.4f}

⚙️  Learning Rate: {logs.get('lr', 0):.6f}
            """)

clean_logger = CleanEpochLogger()

# ========================================
# COMBINE CALLBACKS (WITH NaN DETECTOR)
# ========================================

callbacks = [
    lr_callback,
    nan_detector,  # ← NEW: Stops training if NaN detected
    checkpoint_callback,
    early_stopping,
    tensorboard_callback,
    csv_logger,
    clean_logger
]

print("\n✅ All callbacks configured (including NaN detector)!")

# ========================================
# COMPILE MODEL (FIXED - STRONGER GRADIENT CLIPPING)
# ========================================

print("\n Compiling model...")

optimizer = keras.optimizers.Adam(
    learning_rate=FIXED_LR,  # ← Using fixed lower LR
    clipnorm=5.0,  # ← REDUCED from 10.0 to 5.0 (stronger clipping)
    epsilon=1e-7   # ← Added epsilon for numerical stability
)

model.compile(
    optimizer=optimizer,
    loss=loss_fn,
    metrics=[]
)

print("✅ Model compiled with:")
print(f"   Learning Rate: {FIXED_LR}")
print(f"   Gradient Clipping: 5.0 (stronger than before)")
print(f"   Epsilon: 1e-7 (for numerical stability)")

# ========================================
# TRAINING SUMMARY
# ========================================

print("\n" + "="*70)
print("TRAINING CONFIGURATION (FIXED)")
print("="*70)

print(f"""
Dataset:
   Train: 33,448 images → {33448 // cfg.BATCH_SIZE} steps/epoch
   Valid: 2,300 images → {2300 // cfg.BATCH_SIZE} steps/epoch

Model:
   Parameters: {model.count_params():,}
   Batch Size: {cfg.BATCH_SIZE}
   Epochs: {cfg.EPOCHS}
   Initial LR: {FIXED_LR} (LOWERED from 0.001)

Fixes Applied:
   ✅ Learning rate reduced: 0.001 → 0.0005
   ✅ Gradient clipping strengthened: 10.0 → 5.0
   ✅ NaN detection callback added
   ✅ Numerical stability improved (epsilon)

Saving:
   Checkpoints: {checkpoint_path}
   Logs: {log_dir}
""")

print("="*70)

# ========================================
# START TRAINING
# ========================================

print("\n STARTING TRAINING (WITH FIXES)!")
print("="*70)
print("\n Training will be slower but STABLE")
print(" If NaN still occurs, stop and lower LR to 2e-4")
print("\n" + "="*70 + "\n")

try:
    history = model.fit(
        train_dataset,
        validation_data=valid_dataset,
        epochs=cfg.EPOCHS,
        callbacks=callbacks,
        verbose=1
    )

    print("\n" + "="*70)
    print(" TRAINING COMPLETE!")
    print("="*70)

    # Save final model
    final_model_path = "/content/drive/MyDrive/FYP_Model_Checkpoints/final_model.h5"
    model.save(final_model_path)
    print(f"\n💾 Final model saved: {final_model_path}")

except KeyboardInterrupt:
    print("\n⚠️ Training interrupted by user!")
    print(" Best model saved at checkpoint")

except Exception as e:
    print(f"\n❌ Training error: {e}")
    print(" Check if best model was saved")

    # If NaN error, give specific advice
    if 'nan' in str(e).lower():
        print("\n🚨 NaN Error Detected!")
        print("   SOLUTION: Lower learning rate even more")
        print("   Try: FIXED_LR = 2e-4 (or even 1e-4)")

print("\n✅ Done!")
```
